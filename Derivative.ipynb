The derivative of a function \( f(x) \) at a point \( x = a \) is defined as  

$$
f'(a) = \lim_{h \to 0} \frac{f(a+h) - f(a)}{h}
$$  

This represents the **instantaneous rate of change** of the function at \( x=a \).  

Geometrically, \( f'(a) \) is the slope of the tangent line to the curve \( y = f(x) \) at the point \( (a, f(a)) \).  

- If \( f'(a) > 0 \), the curve is increasing at \( x=a \).  
- If \( f'(a) < 0 \), the curve is decreasing at \( x=a \).  
- If \( f'(a) = 0 \), the curve is flat at that point (possibly a max, min, or saddle point).  

The magnitude \( |f'(a)| \) tells how steep the curve is at that point.  


```python
# We can evaluate the derivative of this function numerically by taking a very small $h$. Of course, the definition asks us to take h->0.

h=0.001
x = 3.0
print('f(x) : ', f(x))

# if x we nudge x in a positive direction, how would the function respond?
# If we move slightly in the positive direction from x (3.0), we can see that the the fucntion is increasing in the graph, and so the result is. 

print('f(x + h) : ', f(x + h)) 

# calculate the slope

## how much the function responded
rise = f(x + h) - f(x)
run = h

slope = rise/run
print('slope : ', slope) 

# This says, at 3 , the slope is ~14
# we can also verify this by differentiating the function, as we do in high school.
# so the differentiation f'(x) = 6x - 4.
# if we plug in x = 3.0, f'(3) = 6(3.0) - 4 = 18-4 = 14






```

```python
# lets calcuate the slope at -3, 

x = -3.0
print('f(x) : ', f(x))

print('f(x + h) : ', f(x + h))

rise = f(x + h) - f(x)
run = h

slope = rise/run

print('slope : ', slope)

# This says, at -3 , the slope is ~-22
# we can also verify this by differentiating the function
# so the differentiation f'(x) = 6x - 4.
# if we plug in x = 3.0, f'(3) = 6(-3.0) - 4 = -18-4 = -22

# The -ve sign indicates that the function decreases. If we slighlty move in positive direction from -3 in the graph, the fucntion goes down.



```

```python
# At some point, the slope will be zero, and this fucntion especially it happens at 2/3.

x = 2/3
print('f(x) : ', f(x))

print('f(x + h) : ', f(x + h))

rise = f(x + h) - f(x)
run = h

slope = rise/run

print('slope : ', slope)

# The slope is 0, which means at that precise point, if we move in positive direction, function doesn't respond, it stays the same almost. Thats why the slope is 0.
```
